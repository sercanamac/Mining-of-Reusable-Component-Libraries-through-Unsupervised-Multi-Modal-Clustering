# Object Description & Viewpoint Viewer

Randomly select 10 objects, show their text descriptions and all 9 rendered viewpoints.

```python
import random
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

random.seed(42)

desc_dir = Path('../../processed_data/descriptions/text')
render_dir = Path('../../processed_data/renders/colored')

VIEWS = ['front', 'front_right', 'right', 'back', 'back_left', 'left', 'top', 'bottom', 'iso']

# Find objects that have both descriptions and renders
desc_gids = {p.stem for p in desc_dir.glob('*.txt')}
render_gids = {p.name for p in render_dir.iterdir() if p.is_dir()}
common_gids = sorted(desc_gids & render_gids)

selected = random.sample(common_gids, 10)
print(f'{len(common_gids)} objects with both descriptions and renders')
print(f'Selected 10 random objects')
```

```python
for gid in selected:
    desc = (desc_dir / f'{gid}.txt').read_text().strip()

    # Collect available views
    view_paths = []
    view_names = []
    for v in VIEWS:
        p = render_dir / gid / f'{v}.png'
        if p.exists():
            view_paths.append(p)
            view_names.append(v)

    n = len(view_paths)
    cols = min(n, 5)
    rows = (n + cols - 1) // cols

    fig, axes = plt.subplots(rows, cols, figsize=(3.5 * cols, 3.5 * rows))
    fig.suptitle(f'{gid}\n{desc}', fontsize=11, fontweight='bold', y=1.02)

    if n == 1:
        axes = [axes]
    else:
        axes = axes.flat if hasattr(axes, 'flat') else [axes]

    for i, ax in enumerate(axes):
        if i < n:
            img = mpimg.imread(str(view_paths[i]))
            ax.imshow(img)
            ax.set_title(view_names[i], fontsize=9)
        ax.axis('off')

    plt.tight_layout()
    plt.show()
```
